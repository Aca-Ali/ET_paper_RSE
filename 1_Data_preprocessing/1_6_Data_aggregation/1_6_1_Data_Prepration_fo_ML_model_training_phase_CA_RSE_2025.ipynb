```python
# training data consist of all the observation from US-UC1 and US-UC2 where ffp flag=0
US_Bi1_ECT_NDVI_ETa_data = pd.concat([US_Bi1_hourly_ffp_NDVI_filtered,
                                       US_Bi1_hourly_ETa_filtered] , axis=1)

US_Bi2_ECT_NDVI_ETa_data = pd.concat([US_Bi2_hourly_ffp_NDVI_filtered,
                                       US_Bi2_hourly_ETa_filtered] , axis=1)

US_Tw3_ECT_NDVI_ETa_data = pd.concat([US_Tw3_hourly_ffp_NDVI_filtered,
                                       US_Tw3_hourly_ETa_filtered] , axis=1)

# Manually remove rows with high uncertainty (this was added after a few rounds of model training/validation) and then add ETa_nan_flag
Bi1_timestamps_to_nan = []

Bi2_timestamps_to_nan = []

Tw3_timestamps_to_nan = []
# addin Nan flag to the data
US_Bi1_ECT_NDVI_ETa_data['ETa_corr'] = US_Bi1_ECT_NDVI_ETa_data['ETa_corr'].mask(US_Bi1_ECT_NDVI_ETa_data.index.isin(Bi1_timestamps_to_nan))
US_Bi1_ECT_NDVI_ETa_data['ETa_NaN_flag'] = US_Bi1_ECT_NDVI_ETa_data['ETa_corr'].isna().astype(int) # add a flag column

US_Bi2_ECT_NDVI_ETa_data['ETa_corr'] = US_Bi2_ECT_NDVI_ETa_data['ETa_corr'].mask(US_Bi2_ECT_NDVI_ETa_data.index.isin(Bi2_timestamps_to_nan))
US_Bi2_ECT_NDVI_ETa_data['ETa_NaN_flag'] = US_Bi2_ECT_NDVI_ETa_data['ETa_corr'].isna().astype(int) # add a flag column

US_Tw3_ECT_NDVI_ETa_data['ETa_corr'] = US_Tw3_ECT_NDVI_ETa_data['ETa_corr'].mask(US_Tw3_ECT_NDVI_ETa_data.index.isin(Tw3_timestamps_to_nan))
US_Tw3_ECT_NDVI_ETa_data['ETa_NaN_flag'] = US_Tw3_ECT_NDVI_ETa_data['ETa_corr'].isna().astype(int) # add a flag column

# rounding up the ETa values to 2 decimal palces
US_Bi1_ECT_NDVI_ETa_data['ETa_corr'] = US_Bi1_ECT_NDVI_ETa_data['ETa_corr']
US_Bi2_ECT_NDVI_ETa_data['ETa_corr'] = US_Bi2_ECT_NDVI_ETa_data['ETa_corr']
```
