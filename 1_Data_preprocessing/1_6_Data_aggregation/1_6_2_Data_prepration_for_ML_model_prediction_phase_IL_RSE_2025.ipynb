```python
def find_nearest_AORC_pixel(target_lat, target_lon, df, lat_col='latitude', lon_col='longitude'):
    """
    Find the nearest coordinate in a DataFrame to a given target latitude and longitude.

    Parameters
    ----------
    df : pd.DataFrame
        DataFrame containing latitude and longitude columns.
    lat_col : str
        Column name for latitude.
    lon_col : str
        Column name for longitude.
    target_lat : float
        Target latitude.
    target_lon : float
        Target longitude.

    Returns
    -------
    pd.DataFrame
        Subset of df with all rows that share the nearest (lat, lon) pair.
    """
    # Ensure numeric type
    df[lat_col] = pd.to_numeric(df[lat_col])
    df[lon_col] = pd.to_numeric(df[lon_col])

    # Get unique coordinate pairs
    unique_coords = df[[lat_col, lon_col]].drop_duplicates().values

    # Find the nearest coordinate pair (extract as float)
    nearest_coords = unique_coords[np.argmin(np.sum((unique_coords - np.asarray([target_lat, target_lon]))**2, axis=1))] 
    nearest_lat, nearest_lon = nearest_coords[0], nearest_coords[1]

    # Keep all rows with that pair
    nearest_rows = df[(df[lat_col] == nearest_lat) & (df[lon_col] == nearest_lon)]

    return nearest_rows
```

```python
def strip_offset_from_index(df: pd.DataFrame) -> pd.DataFrame:
    """
    If DataFrame index is an object dtype and looks like timezone-aware strings
    (e.g. '2020-03-08 01:30:00-07:00'), remove the offset and convert to naive datetimes.
    """
    import re
    df = df.copy()
    
    # Case 1: object index (string timestamps)
    if df.index.dtype == 'O':
        idx_str = df.index.astype(str).str.strip()
        tz_pattern = re.compile(r'(.*?)([+-]\d{2}:\d{2})$')
        
        cleaned = []
        for ts in idx_str:
            match = tz_pattern.match(ts)
            if match:
                cleaned.append(match.group(1).strip())  # keep part before offset
            else:
                cleaned.append(ts)
        
        df.index = pd.to_datetime(cleaned, errors='coerce')

    # Case 2: datetime with timezone
    elif isinstance(df.index, pd.DatetimeIndex) and df.index.tz is not None:
        df.index = df.index.tz_localize(None)
        
    return df
```

```python
import dask.dataframe as dd
import os
import pandas as pd
import numpy as np
from sklearn.neighbors import BallTree
from dask.diagnostics import ProgressBar
from datetime import datetime
import math
import warnings
warnings.filterwarnings('ignore')
```

### Functions

```python
def strip_offset_from_index(df: pd.DataFrame) -> pd.DataFrame:
    """
    If DataFrame index is an object dtype and looks like timezone-aware strings
    (e.g. '2020-03-08 01:30:00-07:00'), remove the offset and convert to naive datetimes.
    """
    import re
    df = df.copy()
    
    # Case 1: object index (string timestamps)
    if df.index.dtype == 'O':
        idx_str = df.index.astype(str).str.strip()
        tz_pattern = re.compile(r'(.*?)([+-]\d{2}:\d{2})$')
        
        cleaned = []
        for ts in idx_str:
            match = tz_pattern.match(ts)
            if match:
                cleaned.append(match.group(1).strip())  # keep part before offset
            else:
                cleaned.append(ts)
        
        df.index = pd.to_datetime(cleaned, errors='coerce')

    # Case 2: datetime with timezone
    elif isinstance(df.index, pd.DatetimeIndex) and df.index.tz is not None:
        df.index = df.index.tz_localize(None)
        
    return df
```

```python
def specific_to_relative_humidity(specific_humidity, temperature, pressure):
    """
    Convert specific humidity to relative humidity.

    Parameters:
    specific_humidity (float): Specific humidity in kg/kg.
    temperature (float): Temperature in degrees Celsius.
    pressure (float): Atmospheric pressure in kPa.

    Returns:
    float: Relative humidity in percentage.
    """

    # Calculate the actual vapor pressure (e)
    e = (specific_humidity * pressure *10 ) / (0.622 + 0.378 * specific_humidity)

    # Calculate the saturation vapor pressure (es) using Tetens' formula
    es = 6.112 * math.exp((17.67 * temperature) / (temperature + 243.5))

    # Calculate relative humidity (RH)
    relative_humidity = (e / es) * 100

    return relative_humidity


def preprocess_aorc_df(raw_AORC_df):

    time_utc = pd.to_datetime(raw_AORC_df['time'], utc=True)
    raw_AORC_df["time_local"] = time_utc.dt.tz_convert("America/Los_Angeles")
    raw_AORC_df.set_index('time_local', inplace=True)
    
    AORC_df_filtered = raw_AORC_df


    # Columns convertion
    AORC_df_filtered['Air Temperature'] = AORC_df_filtered['Air Temperature'] - 273.15

    AORC_df_filtered['Wind Speed'] = np.sqrt(AORC_df_filtered['U-Component of Wind']**2 + AORC_df_filtered['V-Component of Wind']**2)

    AORC_df_filtered['Pressure'] = AORC_df_filtered['Pressure']/1000

    AORC_df_filtered['relative_humidity'] = AORC_df_filtered.apply(
        lambda row: specific_to_relative_humidity(row['Specific Humidity'], row['Air Temperature'], row['Pressure']),
        axis=1
    )

    AORC_df_filtered = AORC_df_filtered.drop(['U-Component of Wind', 'V-Component of Wind', 'Specific Humidity'], axis=1, inplace=False)

    AORC_df_filtered =AORC_df_filtered[['latitude', 'longitude','Total Precipitation', 'Air Temperature','relative_humidity', 
                                              'Downward Long-Wave Radiation Flux', 'Downward Short-Wave Radiation Flux', 'Pressure', 'Wind Speed']]


    return AORC_df_filtered
```

```python
def prepare_AORC_NDVI_df(AORC_df, NDVI_df, date_of_interest, location_name):

    ### Step 0: slice both df to date of interest
    day_NDVI_df = NDVI_df[NDVI_df['Date']==date_of_interest]
    day_AORC_df = AORC_df.loc[datetime.strptime(str(date_of_interest), '%Y%m%d').strftime('%Y-%m-%d')]

    day_NDVI_df['Date'] = pd.to_datetime(day_NDVI_df['Date'].astype(str), format='%Y%m%d').dt.date

    # ---- MET hourly df ----
    day_AORC_df = day_AORC_df.copy()
    day_AORC_df['datetime'] = pd.to_datetime(day_AORC_df.index)
    day_AORC_df['date'] = day_AORC_df['datetime'].dt.date
    day_AORC_df = day_AORC_df.reset_index(drop=True)


    ### Step 1: Precompute nearest coarse pixel (in memory, once)
    coarse_coords = day_AORC_df[['latitude', 'longitude']].drop_duplicates().reset_index(drop=True)

    # build BallTree
    tree = BallTree(np.deg2rad(coarse_coords[['latitude', 'longitude']].values), metric='haversine')

    # query all fine pixels
    fine_rad = np.deg2rad(day_NDVI_df[['latitude', 'longitude']].values)
    dist, idx = tree.query(fine_rad, k=1)

    # attach nearest coarse coords to NDVI df
    day_NDVI_df['lat_coarse'] = coarse_coords.loc[idx[:, 0], 'latitude'].values
    day_NDVI_df['lon_coarse'] = coarse_coords.loc[idx[:, 0], 'longitude'].values

    # Step 2: Find the nearest AORC pixel to each NDVI pixel
    unique_hours = day_AORC_df['datetime'].unique()

    # get the 24 hourly timestamps for that date
    hours_df = day_AORC_df[['datetime', 'date']].drop_duplicates(subset='datetime')

    # add helper key
    day_NDVI_df['key'] = 1
    hours_df['key'] = 1

    # cartesian product (daily × 24 hours)
    df_ndvi_hourly = pd.merge(day_NDVI_df, hours_df, on=['key'], how='left').drop(columns='key')
    df_met_hourly = day_AORC_df

    # Step 3: convert ndvi df to hourly
    # Convert to Dask DataFrames
    ddf_ndvi = dd.from_pandas(df_ndvi_hourly, npartitions=32)
    ddf_met = dd.from_pandas(df_met_hourly, npartitions=1)

    # Step 4: now add information from AORC df 
    ddf_merged = ddf_ndvi.merge(
        ddf_met,
        left_on=['datetime', 'lat_coarse', 'lon_coarse'],
        right_on=['datetime', 'latitude', 'longitude'],
        how='left',
        suffixes=('', '_met')
    )


    with ProgressBar():
        ddf_merged.to_parquet(
            f"{location_name}_{date_of_interest}_ndvi_hourly_with_met/",
            write_index=False,
            compression='snappy',
            engine='pyarrow'
        )
    
```
