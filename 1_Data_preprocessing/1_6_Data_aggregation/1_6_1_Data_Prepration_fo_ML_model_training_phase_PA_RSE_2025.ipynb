```python
# hourly ETa from EC tower
USUC1_df_FilteredDates_2019_ETa = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '1_Initial_data_preprocessing', 'USUC1_df_2019_ETa_MET.csv'),
                                             index_col = 'TIMESTAMP_START', parse_dates=['TIMESTAMP_START'])
USUC1_df_FilteredDates_2020_ETa = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '1_Initial_data_preprocessing', 'USUC1_df_2020_ETa_MET.csv'),
                                             index_col = 'TIMESTAMP_START', parse_dates=['TIMESTAMP_START'])
USUC1_df_FilteredDates_2021_ETa = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '1_Initial_data_preprocessing', 'USUC1_df_2021_ETa_MET.csv'),
                                             index_col = 'TIMESTAMP_START', parse_dates=['TIMESTAMP_START'])
USUC1_df_FilteredDates_2022_ETa = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '1_Initial_data_preprocessing', 'USUC1_df_2022_ETa_MET.csv'),
                                             index_col = 'TIMESTAMP_START', parse_dates=['TIMESTAMP_START'])

USUC1_df_FilteredDates_2023_ETa = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '1_Initial_data_preprocessing', 'USUC1_df_2023_ETa_MET.csv'),
                                             index_col = 'TIMESTAMP_START', parse_dates=['TIMESTAMP_START'])
USUC1_df_FilteredDates_2024_ETa = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '1_Initial_data_preprocessing', 'USUC1_df_2024_ETa_MET.csv'),
                                             index_col = 'TIMESTAMP_START', parse_dates=['TIMESTAMP_START'])

USUC2_df_FilteredDates_2019_ETa = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '1_Initial_data_preprocessing', 'USUC2_df_2019_ETa_MET.csv'),
                                             index_col = 'TIMESTAMP_START', parse_dates=['TIMESTAMP_START'])
USUC2_df_FilteredDates_2020_ETa = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '1_Initial_data_preprocessing', 'USUC2_df_2020_ETa_MET.csv'),
                                             index_col = 'TIMESTAMP_START', parse_dates=['TIMESTAMP_START'])
USUC2_df_FilteredDates_2021_ETa = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '1_Initial_data_preprocessing', 'USUC2_df_2021_ETa_MET.csv'),
                                             index_col = 'TIMESTAMP_START', parse_dates=['TIMESTAMP_START'])
USUC2_df_FilteredDates_2022_ETa = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '1_Initial_data_preprocessing', 'USUC2_df_2022_ETa_MET.csv'),
                                             index_col = 'TIMESTAMP_START', parse_dates=['TIMESTAMP_START'])

USUC2_df_FilteredDates_2023_ETa = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '1_Initial_data_preprocessing', 'USUC2_df_2023_ETa_MET.csv'),
                                             index_col = 'TIMESTAMP_START', parse_dates=['TIMESTAMP_START'])
USUC2_df_FilteredDates_2024_ETa = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '1_Initial_data_preprocessing', 'USUC2_df_2024_ETa_MET.csv'),
                                             index_col = 'TIMESTAMP_START', parse_dates=['TIMESTAMP_START'])

USHWB_df_FilteredDates_2017_ETa = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '1_Initial_data_preprocessing', 'USHWB_df_2017_ETa_MET.csv'),
                                             index_col = 'TIMESTAMP_START', parse_dates=['TIMESTAMP_START'])


# Concat all the years
US_UC1_hourly_ETa_filtered = pd.concat([USUC1_df_FilteredDates_2019_ETa, USUC1_df_FilteredDates_2020_ETa,
                                        USUC1_df_FilteredDates_2021_ETa, USUC1_df_FilteredDates_2022_ETa,
                                        USUC1_df_FilteredDates_2023_ETa, USUC1_df_FilteredDates_2024_ETa], axis=0)
US_UC2_hourly_ETa_filtered = pd.concat([USUC2_df_FilteredDates_2019_ETa, USUC2_df_FilteredDates_2020_ETa,
                                        USUC2_df_FilteredDates_2021_ETa, USUC2_df_FilteredDates_2022_ETa,
                                        USUC2_df_FilteredDates_2023_ETa, USUC2_df_FilteredDates_2024_ETa], axis=0)
```

```python
# median FFP NDVI 
US_UC1_2019_hourly_ffp_NDVI = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '3_PlanetScope','PS_AVE_Hourly_FFP_NDVI_UC1_2019.csv'),
                                          parse_dates=[['Date', 'hour']]).set_index('Date_hour', inplace=False)
US_UC1_2020_hourly_ffp_NDVI = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '3_PlanetScope', 'PS_AVE_Hourly_FFP_NDVI_UC1_2020.csv'),
                                          parse_dates=[['Date', 'hour']]).set_index('Date_hour', inplace=False)
US_UC1_2021_hourly_ffp_NDVI = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '3_PlanetScope', 'PS_AVE_Hourly_FFP_NDVI_UC1_2021.csv'),
                                          parse_dates=[['Date', 'hour']]).set_index('Date_hour', inplace=False)
US_UC1_2022_hourly_ffp_NDVI = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '3_PlanetScope', 'PS_AVE_Hourly_FFP_NDVI_UC1_2022.csv'),
                                          parse_dates=[['Date', 'hour']]).set_index('Date_hour', inplace=False)
US_UC1_2023_hourly_ffp_NDVI = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '3_PlanetScope', 'PS_AVE_Hourly_FFP_NDVI_UC1_2023.csv'),
                                          parse_dates=[['Date', 'hour']]).set_index('Date_hour', inplace=False)
US_UC1_2024_hourly_ffp_NDVI = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '3_PlanetScope', 'PS_AVE_Hourly_FFP_NDVI_UC1_2024.csv'),
                                          parse_dates=[['Date', 'hour']]).set_index('Date_hour', inplace=False)
                                                       
US_UC2_2019_hourly_ffp_NDVI = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '3_PlanetScope', 'PS_AVE_Hourly_FFP_NDVI_UC2_2019.csv'),
                                          parse_dates=[['Date', 'hour']]).set_index('Date_hour', inplace=False)
US_UC2_2020_hourly_ffp_NDVI = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '3_PlanetScope', 'PS_AVE_Hourly_FFP_NDVI_UC2_2020.csv'),
                                          parse_dates=[['Date', 'hour']]).set_index('Date_hour', inplace=False)
US_UC2_2021_hourly_ffp_NDVI = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '3_PlanetScope', 'PS_AVE_Hourly_FFP_NDVI_UC2_2021.csv'),
                                          parse_dates=[['Date', 'hour']]).set_index('Date_hour', inplace=False)
US_UC2_2022_hourly_ffp_NDVI = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '3_PlanetScope', 'PS_AVE_Hourly_FFP_NDVI_UC2_2022.csv'),
                                          parse_dates=[['Date', 'hour']]).set_index('Date_hour', inplace=False)
US_UC2_2023_hourly_ffp_NDVI = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '3_PlanetScope', 'PS_AVE_Hourly_FFP_NDVI_UC2_2023.csv'),
                                          parse_dates=[['Date', 'hour']]).set_index('Date_hour', inplace=False)
US_UC2_2024_hourly_ffp_NDVI = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '3_PlanetScope', 'PS_AVE_Hourly_FFP_NDVI_UC2_2024.csv'),
                                          parse_dates=[['Date', 'hour']]).set_index('Date_hour', inplace=False)

US_HWB_2017_hourly_ffp_NDVI = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), '3_PlanetScope', 'PS_AVE_Hourly_FFP_NDVI_HWB_2017.csv'),
                                          parse_dates=[['Date', 'hour']]).set_index('Date_hour', inplace=False)


# Concat all the years
US_UC1_hourly_ffp_NDVI_filtered = pd.concat([US_UC1_2019_hourly_ffp_NDVI, US_UC1_2020_hourly_ffp_NDVI,
                                             US_UC1_2021_hourly_ffp_NDVI, US_UC1_2022_hourly_ffp_NDVI,
                                             US_UC1_2023_hourly_ffp_NDVI, US_UC1_2024_hourly_ffp_NDVI], axis=0)

US_UC2_hourly_ffp_NDVI_filtered = pd.concat([US_UC2_2019_hourly_ffp_NDVI, US_UC2_2020_hourly_ffp_NDVI,
                                             US_UC2_2021_hourly_ffp_NDVI, US_UC2_2022_hourly_ffp_NDVI,
                                             US_UC2_2023_hourly_ffp_NDVI, US_UC2_2024_hourly_ffp_NDVI], axis=0)
```

```python
# training data consist of all the observation from US-UC1 and US-UC2 where ffp flag=0
US_UC1_ECT_NDVI_ETa_data = pd.concat([US_UC1_hourly_ffp_NDVI_filtered,
                                       US_UC1_hourly_ETa_filtered] , axis=1)

US_UC2_ECT_NDVI_ETa_data = pd.concat([US_UC2_hourly_ffp_NDVI_filtered,
                                       US_UC2_hourly_ETa_filtered] , axis=1)

US_HWB_ECT_NDVI_ETa_data = pd.concat([USHWB_df_FilteredDates_2017_ETa,
                                       US_HWB_2017_hourly_ffp_NDVI] , axis=1)

# Manually remove rows with high uncertainty (this was added after a few rounds of model training/validation) and then add ETa_nan_flag
UC1_timestamps_to_nan = [pd.Timestamp("2019-06-10 8:00:00"),pd.Timestamp("2019-05-20 12:00:00"),
                         pd.Timestamp("2021-06-30 13:00:00"), pd.Timestamp("2021-06-30 11:00:00"),
                         pd.Timestamp("2021-09-10 13:00:00"),
                         pd.Timestamp("2022-07-03 12:00:00"), pd.Timestamp("2022-07-25 12:00:00")]

UC2_timestamps_to_nan = [pd.Timestamp("2020-7-11 12:00:00"), pd.Timestamp("2020-7-11 13:00:00"),
                         pd.Timestamp("2021-10-16 09:00:00"), pd.Timestamp("2021-10-31 09:00:00"), pd.Timestamp("2021-06-30 13:00:00"),
                         pd.Timestamp("2021-07-26 12:00:00"), pd.Timestamp("2021-07-15 13:00:00"), pd.Timestamp("2021-06-30 11:00:00"),
                         pd.Timestamp("2022-08-07 13:00:00"), pd.Timestamp("2022-11-13 14:00:00"),
                         pd.Timestamp("2022-12-03 13:00:00"), pd.Timestamp("2022-12-03 14:00:00"), pd.Timestamp("2022-07-25 12:00:00")]

HWB_timestamps_to_nan = []
# addin Nan flag to the data
US_UC1_ECT_NDVI_ETa_data['ETa_corr'] = US_UC1_ECT_NDVI_ETa_data['ETa_corr'].mask(US_UC1_ECT_NDVI_ETa_data.index.isin(UC1_timestamps_to_nan))
US_UC1_ECT_NDVI_ETa_data['ETa_NaN_flag'] = US_UC1_ECT_NDVI_ETa_data['ETa_corr'].isna().astype(int) # add a flag column

US_UC2_ECT_NDVI_ETa_data['ETa_corr'] = US_UC2_ECT_NDVI_ETa_data['ETa_corr'].mask(US_UC2_ECT_NDVI_ETa_data.index.isin(UC2_timestamps_to_nan))
US_UC2_ECT_NDVI_ETa_data['ETa_NaN_flag'] = US_UC2_ECT_NDVI_ETa_data['ETa_corr'].isna().astype(int) # add a flag column

US_HWB_ECT_NDVI_ETa_data['ETa_corr'] = US_HWB_ECT_NDVI_ETa_data['ETa_corr'].mask(US_HWB_ECT_NDVI_ETa_data.index.isin(HWB_timestamps_to_nan))
US_HWB_ECT_NDVI_ETa_data['ETa_NaN_flag'] = US_HWB_ECT_NDVI_ETa_data['ETa_corr'].isna().astype(int) # add a flag column

# rounding up the ETa values to 2 decimal palces
US_UC1_ECT_NDVI_ETa_data['ETa_corr'] = US_UC1_ECT_NDVI_ETa_data['ETa_corr'].round(2)
US_UC2_ECT_NDVI_ETa_data['ETa_corr'] = US_UC2_ECT_NDVI_ETa_data['ETa_corr'].round(2)
US_HWB_ECT_NDVI_ETa_data['ETa_corr'] = US_HWB_ECT_NDVI_ETa_data['ETa_corr'].round(2)
```

```python
US_UC1_ECT_NDVI_ETa_data.to_csv('US_UC1_Hourly_ECT_PSNDVI_ETa.csv')
US_UC2_ECT_NDVI_ETa_data.to_csv('US_UC2_Hourly_ECT_PSNDVI_ETa.csv')
US_HWB_ECT_NDVI_ETa_data.to_csv('US_HWB_Hourly_ECT_PSNDVI_ETa.csv')
```
